## 1. Imports

```python
import pydara
from pydara import tse
print(pydara.__version__)
```

## 2. Helper

```python
def show(frame):
    print("Rows:", len(frame))
    display(tse.select_columns(frame))
    print(tse.frame_metadata(frame))
```

## 3. Explicit network permission

```python
SYMBOL = "فولاد"
INDEX = "32097828799138957"
RUN_LIVE = False
```

## 4. Current watch

```python
if RUN_LIVE:
    watch = tse.get_market_watch()
    show(watch)
else:
    print("Live calls disabled. Set RUN_LIVE=True after reviewing the guide.")
```

## 5. Natural/legal history

```python
if RUN_LIVE:
    people = tse.get_client_type_history(SYMBOL, include_jdate=True)
    show(people)
```

## 6. Inspect index identities first

```python
if RUN_LIVE:
    indices = tse.get_indices()
    show(indices)
```

## 7. Index history

```python
if RUN_LIVE:
    history = tse.get_index_history(INDEX, include_jdate=True)
    show(history)
```

## 8. Explicitly finish queued writes

```python
if RUN_LIVE:
    print(tse.flush(timeout=10, raise_on_error=False))
    print(tse.close())
```

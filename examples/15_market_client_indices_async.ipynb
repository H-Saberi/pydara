## 1. Imports

```python
from pydara import tse
from pydara.tse.samples.market import sample_market_async_client
```

## 2. Settings

```python
SYMBOL = "فولاد"
INDEX = "32097828799138957"
```

## 3. Native notebook await

```python
async with sample_market_async_client() as client:
    people = await client.get_client_type_history(SYMBOL)
    history = await client.get_index_history(INDEX)
    watch = await client.get_market_watch()
display(tse.select_columns(people))
display(tse.select_columns(history))
display(tse.select_columns(watch))
```

## 4. Bounded polling, synthetic only

```python
async with sample_market_async_client() as client:
    async for snapshot in client.watch_market(interval=1, max_updates=2):
        print("Snapshot", snapshot.attrs["pydara"]["poll_index"], "rows", len(snapshot))
```

## 1 - ورود کتابخانه‌ها

```python
import pandas as pd
import pydara
from pydara import tse
from pathlib import Path
from tempfile import TemporaryDirectory
from pydara.tse.samples.extended import sample_extended_client
print(pydara.__version__)
assert pydara.__version__ == "0.7.3a6"
```

## 2 - تعریف توابع

```python
def show(frame):
    display(tse.select_columns(frame).head())
    return len(frame)
```

## 3 - تنظیمات

```python
SYMBOL = 'فولاد'
START = '1403-12-27'
END = '1403-12-28'
```

## 4 - اجرا

```python
with sample_extended_client() as client:
    trades = client.get_trades(SYMBOL, day='2025-03-18')
    books = client.get_order_book_history(SYMBOL, start='2025-03-18')
    adjustments = client.get_adjustments(SYMBOL)
    adjusted = client.get_history(SYMBOL, adjustment='source')
    holders = client.get_shareholders(SYMBOL)
    nav = client.get_fund_nav_history('7001')
    options = client.get_options(SYMBOL)
    for frame in (trades, books, adjustments, adjusted, holders, nav, options):
        show(frame)
        display(tse.validate_data(frame))
```

## 5 - گزارش نتیجه

```python
assert trades.trade_id.nunique() == 2
assert books.source_position.nunique() == 2
assert options.option_type.nunique() == 2
print('Unknown fields were not invented. Raw history stays separate from adjusted.')
```

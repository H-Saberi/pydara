```python
# سلول ۱ — کتابخانه‌ها
from pydara import tse
import pandas as pd
```

```python
# سلول ۲ — توابع نمایش و کنترل
def show_report(data):
    display(data.head())
    print(data.attrs.get("pydara", {}).get("errors", []))
```

```python
# سلول ۳ — تنظیمات
SELECTED_MARKETS = ["TSE", "IFB"]
SECTOR = None  # Optional industry code from the real source
```

```python
# سلول ۴ — دریافت داده‌ها (به اینترنت یا کش نیاز دارد)
markets_df = tse.markets(market=SELECTED_MARKETS)
assets_df = tse.assets(markets=markets_df, **({"industry_code": SECTOR} if SECTOR else {}))
prices_df = tse.prices(markets=markets_df)
orders_df = tse.orders(assets=assets_df.head(3))
```

```python
# سلول ۵ — گزارش و خروجی
show_report(markets_df)
show_report(assets_df)
show_report(prices_df)
show_report(orders_df)
# prices_df.to_csv("daily_prices.csv", index=False)
```

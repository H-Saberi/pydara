# رابط Async تاریخچه — نمونه مصنوعی

این دفترچه از حلقه رویداد خود Jupyter استفاده می‌کند و `asyncio.run()` فراخوانی نمی‌شود. هیچ قیمت یا داده‌ای در این نمونه واقعی نیست.

## سلول ۱ — ورود کتابخانه‌ها

```python
import pydara
from pydara.tse.samples import sample_async_client
print(pydara.__version__)
```

## سلول ۲ — تعریف تابع

```python
async def receive_examples(symbol, symbols):
    async with sample_async_client() as client:
        frame = await client.get_history(symbol, include_jdate=True)
        batch = await client.get_history_many(symbols, on_error="collect")
        chunks = [part async for part in client.iter_history(
            symbol, start="2024-03-19", end="2025-03-21", chunk_days=200
        )]
    return frame, batch, chunks
```

## سلول ۳ — تنظیمات

```python
SYMBOL = "فولاد"
SYMBOLS = ["فولاد", "شستا", "نمادناموجود"]
```

## سلول ۴ — اجرا

```python
frame, batch, chunks = await receive_examples(SYMBOL, SYMBOLS)
assert len(frame) == 6 and len(batch.data) == 12 and len(batch.errors) == 1
```

## سلول ۵ — نمایش

```python
display(frame[["date", "jdate", "last_price", "closing_price"]])
display(batch.requests)
print([len(chunk) for chunk in chunks])
```

برای دریافت واقعی، `sample_async_client()` را با `AsyncClient()` جایگزین کن. این تغییر نیازمند اینترنت است؛ مدل DataFrame و سیاست خطا همان خواهد بود.

## 1 - ورود کتابخانه‌ها

```python
import pydara
from pydara import tse
from pydara.tse import Client, ClientConfig, NetworkConfig
print(pydara.__version__)
```

## 2 - تعریف توابع

```python
def show(frame):
    display(tse.select_columns(frame).head())
```

## 3 - تنظیمات

```python
RUN_LIVE = False
DIRECT = True  # Explicit choice; TLS is still verified.
SYMBOL = 'فولاد'
```

## 4 - اجرا

```python
if RUN_LIVE:
    config = ClientConfig(network=NetworkConfig(trust_env=not DIRECT, max_retries=0, request_budget=20))
    with Client(config=config, storage=False) as client:
        show(client.get_quote_frame(SYMBOL))
        show(client.get_order_book(SYMBOL))
        show(client.get_trades(SYMBOL))
        show(client.get_shareholders(SYMBOL))
        show(client.get_options(SYMBOL))
else:
    print('No network request: RUN_LIVE=False')
```
